```python
players = {
    "Serena Williams": {"sport": "Tennis", "achievements": 23},
    "Lionel Messi": {"sport": "Soccer", "achievements": 7},
    "Michael Phelps": {"sport": "Swimming", "achievements": 23},
    "Usain Bolt": {"sport": "Athletics", "achievements": 8},
    "Roger Federer": {"sport": "Tennis", "achievements": 20},
    "Cristiano Ronaldo": {"sport": "Soccer", "achievements": 5}
}

player = "Lionel Messi"
data = players[player]

if data["achievements"] > 10:
    print(player, data["sport"], data["achievements"])
else:
    print("does not have more than 10 achievements")
```

```python
num = int(input("Enter an integer between 1 and 9999: "))

if num % 2 == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")

str_num = str(num)
if str_num == str_num[::-1]:
    print(f"{num} is a palindrome")
else:
    print(f"{num} is not a palindrome")

digit_sum = sum(int(d) for d in str_num)
if num % digit_sum == 0:
    print(f"{num} is divisible by the sum of its digits ({digit_sum})")
else:
    print(f"{num} is not divisible by the sum of its digits ({digit_sum})")
```

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']

for color in squares:
    print(color)
```

```python
new_squares = []
i = 0
while i < len(squares):
    if squares[i] == 'orange':
        new_squares.append(squares[i])
        i += 1
    else:
        break

print(new_squares)
```

```python
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]
seven_letter_animals = []

i = 0
while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])
    i += 1

print(seven_letter_animals)
```
