```python
prices = (250, 300, 150, 400, 100, 350, 200)
```

```python
print("Highest price:", max(prices))
print("Lowest price:", min(prices))
print("Total sum of prices:", sum(prices))
```

```python
sorted_prices = sorted(prices)
print(type(sorted_prices))
print("Sorted prices (list):", sorted_prices)
```

```python
try:
    prices[0] = 500
except TypeError as err:
    print("Error while modifying tuple:", err)
    print("Explanation: Tuples are immutable, so their elements cannot be changed after creation.")
```
